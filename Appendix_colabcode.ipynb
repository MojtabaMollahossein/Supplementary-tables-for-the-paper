```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Data
data = {
    "Prediction horizon": [
        "One-step", "One-step", "One-step",
        "50-step", "50-step", "50-step"
    ],
    "Model / strategy": [
        "SIC",
        "DIC from scratch",
        "Proposed staged DIC",
        "SIC",
        "DIC from scratch",
        "Proposed staged DIC"
    ],
    "Evaluation NRMSE": [0.054, 0.051, 0.026, 0.145, 0.125, 0.069],
    "Evaluation NMAE":  [0.037, 0.035, 0.020, 0.101, 0.0867, 0.046],
    "Evaluation R²":    [0.954, 0.958, 0.989, 0.661, 0.729, 0.924]
}

df = pd.DataFrame(data)

# Display clean table
display(df)

# Print markdown table with 3 decimals
print("📊 Comparison of SIC, DIC from scratch, and proposed staged DIC:")
print(df.to_markdown(index=False, floatfmt=".3f"))


# Function for clean grouped bar plots
def plot_metric(metric_name):
    pivot_df = df.pivot(
        index="Prediction horizon",
        columns="Model / strategy",
        values=metric_name
    )

    ax = pivot_df.plot(kind="bar", figsize=(8, 5), edgecolor="black")

    ax.set_title(metric_name, fontsize=14, fontweight="bold")
    ax.set_xlabel("Prediction horizon", fontsize=12)
    ax.set_ylabel(metric_name, fontsize=12)
    ax.grid(axis="y", linestyle="--", alpha=0.5)
    ax.legend(title="Model / strategy", fontsize=9)
    ax.tick_params(axis="x", rotation=0)

    # Add values above bars
    for container in ax.containers:
        ax.bar_label(container, fmt="%.3f", fontsize=9, padding=3)

    plt.tight_layout()
    plt.show()


# Plot each metric separately
plot_metric("Evaluation NRMSE")
plot_metric("Evaluation NMAE")
plot_metric("Evaluation R²")
```

To minimize the number of tables in the paper, only the results for one-step prediction in the SIC scenario are presented. Tables 11 and 12 demonstrate the robustness of the transfer-learning model when applied to new subjects.
